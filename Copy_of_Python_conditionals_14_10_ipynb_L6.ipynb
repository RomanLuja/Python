## 1. `if` / `else` — 3 завдання

### 📘 Теорія: `if` і `else`

> 📌 **Шаблон синтаксису** — це схема, а не код для запуску
>
> <code>if condition:<br>&nbsp;&nbsp;&nbsp;&nbsp;# виконується, якщо condition істинна (True)<br>else:<br>&nbsp;&nbsp;&nbsp;&nbsp;# виконується, якщо condition хибна (False)</code>

**Правила синтаксису:**
- після умови й після `else` обовʼязково ставиться **двокрапка** `:`;
- код усередині гілки пишеться з **відступом у 4 пробіли** — саме відступ показує Python, які рядки належать до умови;
- усі рядки однієї гілки мають однаковий відступ; рядок без відступу вже не належить до умови й виконується завжди;
- `else` не має власної умови й не є обовʼязковим.

> ⚠️ Неправильний відступ — найчастіша помилка новачків. Python повідомить про неї як `IndentationError`.

```python
balance = 1200
price = 950

if balance >= price:
    print("Payment successful")
    balance = balance - price
else:
    print("Insufficient funds")

print("Balance:", balance)   # цей рядок без відступу — виконується завжди

# Результат:
# Payment successful
# Balance: 250
```

### 📘 Істинність значень (truthiness)

В умові може стояти не лише порівняння, а **будь-яке значення**. Python сам перетворить його на `True` чи `False`.

| Вважаються хибними (`False`) | Усе інше — істинне (`True`) |
|---|---|
| `False`, `None` | `True` |
| `0`, `0.0` | будь-яке ненульове число, навіть відʼємне |
| порожній рядок `""` | непорожній рядок, навіть `" "` |
| порожні колекції: `[]`, `()`, `{}`, `set()` | колекція хоча б з одним елементом |

Тому замість `if len(cart) > 0:` зазвичай пишуть просто `if cart:`.

```python
cart = []
if cart:
    print("Items in cart:", len(cart))
else:
    print("Cart is empty")       # Cart is empty

comment = "Great service!"
if comment:
    print("Feedback received")   # Feedback received

print(bool(0), bool(""), bool([]), bool(None))   # False False False False
print(bool(-5), bool(" "), bool([0]))            # True True True
```

## 2. `elif` — 4 завдання

### 📘 Теорія: `elif`

Коли варіантів більше ніж два, використовують `elif` (скорочення від *else if*):

> 📌 **Шаблон синтаксису** — це схема, а не код для запуску
>
> <code>if condition_1:<br>&nbsp;&nbsp;&nbsp;&nbsp;# варіант 1<br>elif condition_2:<br>&nbsp;&nbsp;&nbsp;&nbsp;# варіант 2<br>elif condition_3:<br>&nbsp;&nbsp;&nbsp;&nbsp;# варіант 3<br>else:<br>&nbsp;&nbsp;&nbsp;&nbsp;# жодна умова не виконалася</code>

**Як це працює:**
- умови перевіряються **зверху вниз**;
- виконується **лише перша** гілка, умова якої істинна; решта пропускаються, навіть якщо вони теж істинні;
- `elif` може бути скільки завгодно, `else` — не обовʼязковий.

> 💡 **Порядок має значення.** Для порогових значень («від 90», «від 75», «від 60») починайте з найсуворішої умови.

```python
score = 78

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(grade)   # B
```

### ⚠️ Дві типові помилки з `elif`

```python
score = 95

# ❌ Неправильний порядок: перша ж істинна умова «забирає» результат
if score >= 60:
    print("C")
elif score >= 90:
    print("A")      # сюди програма ніколи не дійде
# Виведе: C

# ❌ Окремі if замість elif: спрацюють УСІ істинні умови
if score >= 90:
    print("A")
if score >= 60:
    print("C")
# Виведе: A, а потім C
```

## 3. Логічні оператори в умовах — 5 завдань

### 📘 Теорія: `and`, `or`, `not`

Логічні оператори поєднують кілька умов в одну.

| Оператор | Значення | Результат `True`, коли… |
|---|---|---|
| `a and b` | і | **обидві** умови істинні |
| `a or b` | або | **хоча б одна** умова істинна |
| `not a` | не | умова хибна (інвертує значення) |

**Таблиця істинності:**

| `a` | `b` | `a and b` | `a or b` | `not a` |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

**Пріоритет:** спочатку виконується `not`, потім `and`, потім `or`. Якщо сумніваєтеся — ставте дужки: вони роблять код зрозумілішим.

**Корисні прийоми:**
- **ланцюжок порівнянь:** `18 <= age <= 65` — те саме, що `age >= 18 and age <= 65`;
- **`in` / `not in`** перевіряють наявність елемента у списку, кортежі, множині, ключа в словнику чи підрядка в рядку.

```python
age = 25
has_ticket = True
city = "Lviv"

print(age >= 18 and has_ticket)       # True
print(age < 18 or not has_ticket)     # False
print(18 <= age <= 30)                # True — ланцюжок порівнянь

if age >= 18 and has_ticket:
    print("Entry allowed")            # Entry allowed

if city in ["Kyiv", "Lviv", "Odesa"]:
    print("Free delivery")            # Free delivery

blocked = {"spam@mail.com", "fake@mail.com"}
email = "anna@mail.com"
if email not in blocked:
    print("Email accepted")           # Email accepted
```

**Пріоритет операторів на прикладі:**

```python
is_vip = True
amount = 300
has_coupon = False

# and виконується раніше за or: is_vip or (amount > 500 and has_coupon)
print(is_vip or amount > 500 and has_coupon)     # True

# дужки змінюють порядок обчислення
print((is_vip or amount > 500) and has_coupon)   # False
```

### ⚠️ Типові помилки в умовах

| `if x = 5:` | `SyntaxError`: `=` — це присвоєння | `if x == 5:` |

## 4. Вкладені умови — 3 завдання

### 📘 Теорія: вкладені умови

Усередині гілки `if`, `elif` чи `else` можна написати ще одну умову. Кожен рівень вкладеності — це ще один відступ у 4 пробіли.

**Коли вкладені умови доречні:**
- друга перевірка має сенс лише після того, як пройшла перша (спочатку перевіряємо, чи клієнт зареєстрований, і лише потім — його статус);
- у різних гілках потрібні **різні** додаткові перевірки.

**Коли краще обійтися без них:**
- якщо потрібна лише одна гілка «так», вкладеність замінюється на `and`;
- якщо вкладені умови перетворилися на «драбину» з 3–4 рівнів — спробуйте переписати через `elif` та логічні оператори. Плаский код легше читати й перевіряти.

```python
is_member = True
amount = 1800

if is_member:
    if amount >= 1500:
        discount = 15
    else:
        discount = 5
else:
    if amount >= 2000:
        discount = 7
    else:
        discount = 0

print("Discount:", discount, "%")   # Discount: 15 %
```

```python
age = 20
has_license = True

# вкладені умови: у кожної гілки — своє повідомлення
if age >= 18:
    if has_license:
        print("Can rent a car")
    else:
        print("Need a driving license")
else:
    print("Too young")

# якщо потрібна лише гілка «так» — вкладеність замінюємо на and
if age >= 18 and has_license:
    print("Can rent a car")
```

## 5. `match` / `case` — 3 завдання

### 📘 Теорія: `match` / `case`

Конструкція `match` / `case` (зʼявилася в Python 3.10) порівнює одне значення з кількома **шаблонами** і виконує гілку першого шаблону, що збігся. Вона зручна, коли одну змінну треба порівняти з багатьма конкретними значеннями.

> 📌 **Шаблон синтаксису** — це схема, а не код для запуску
>
> <code>match value:<br>&nbsp;&nbsp;&nbsp;&nbsp;case pattern_1:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;# дії<br>&nbsp;&nbsp;&nbsp;&nbsp;case pattern_2:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;# дії<br>&nbsp;&nbsp;&nbsp;&nbsp;case _:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;# жоден шаблон не збігся (аналог else)</code>

| Шаблон | Що означає |
|---|---|
| `case "paid":` | значення дорівнює `"paid"` |
| <code>case "new" &#124; "pending":</code> | значення дорівнює **одному з** перелічених |
| `case _:` | будь-яке значення — гілка «за замовчуванням», завжди ставиться останньою |
| `case (0, y):` | кортеж з двох елементів, перший дорівнює 0; другий записується у змінну `y` |

**Важливо:**
- виконується **лише одна** гілка — перша, що збіглася;
- `match` найкраще працює для **точних значень**. Для діапазонів («більше 100», «від 18 до 65») зручніше `if` / `elif`.

```python
status = "shipped"

match status:
    case "new":
        print("Order received")
    case "paid":
        print("Waiting for shipment")
    case "shipped" | "in transit":
        print("On the way")
    case "delivered":
        print("Delivered")
    case _:
        print("Unknown status")

# On the way
```

**Шаблони для кортежів.** `match` вміє не лише порівнювати, а й «розбирати» структуру: перевіряти кількість елементів, конкретні значення і записувати решту в змінні.

```python
point = (0, 5)

match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print("On Y axis, y =", y)
    case (x, 0):
        print("On X axis, x =", x)
    case _:
        print("Somewhere else")

# On Y axis, y = 5
```

## 6. Умовний вираз (тернарний оператор) — 3 завдання

### 📘 Теорія: умовний вираз

Коли треба просто **вибрати одне з двох значень**, чотири рядки `if` / `else` можна записати в один:

> 📌 **Шаблон синтаксису** — це схема, а не код для запуску
>
> <code>value = value_if_true if condition else value_if_false</code>

Читається майже як звичайна мова: «значення А, **якщо** умова, **інакше** значення Б».

Умовний вираз **повертає значення**, тому його можна:
- записати у змінну;
- передати у `print()`;
- використати всередині обчислень.

**Коли використовувати:** простий вибір між двома значеннями.  
**Коли не варто:** якщо варіантів більше двох, або в гілках треба виконати кілька дій — тоді звичайний `if` / `elif` / `else` читається значно краще.

```python
age = 16

# звичайний if / else
if age >= 18:
    status = "adult"
else:
    status = "minor"

# те саме — умовним виразом
status = "adult" if age >= 18 else "minor"
print(status)                                          # minor

stock = 0
print("In stock" if stock > 0 else "Out of stock")    # Out of stock

price = 1200
is_vip = True
final_price = price * 0.9 if is_vip else price
print(final_price)                                     # 1080.0
```

```python
score = 82

# ❌ працює, але читати такий код важко
grade = "A" if score >= 90 else "B" if score >= 75 else "C"
print(grade)   # B

# ✅ для трьох і більше варіантів краще if / elif / else
```

## 🧭 Підсумок: яку конструкцію обрати?

| Ситуація | Конструкція |
|---|---|
| Виконати дію лише за певної умови | `if` |
| Два варіанти дій | `if` / `else` |
| Три й більше варіантів, діапазони значень | `if` / `elif` / `else` |
| Кілька умов одночасно | `and`, `or`, `not`, `in` |
| Друга перевірка має сенс лише після першої | вкладені умови |
| Одна змінна порівнюється з багатьма конкретними значеннями | `match` / `case` |
| Вибір одного з двох значень | `value_a if condition else value_b` |

**Чек-лист перед запуском коду з умовами:**
- після умови стоїть двокрапка `:`;
- код усередині гілки має відступ у 4 пробіли;
- для порівняння використано `==`, а не `=`;
- у кожному порівнянні з `or` / `and` повністю записана ліва частина (`x == 1 or x == 2`, а не `x == 1 or 2`);
- у ланцюжку `elif` умови стоять у правильному порядку;
- програму перевірено на кількох різних значеннях, включно з граничними (рівно 1000, рівно 18 тощо).
