## 1.1 Навіщо потрібні функції

**Функція** — іменований блок коду, що виконує одну задачу. Логіку пишемо один раз, а використовуємо скільки завгодно. Це принцип **DRY** (*Don't Repeat Yourself*): якщо формула зміниться, правити доведеться лише в одному місці.

````python
# Без функції: однаковий розрахунок повторюється
revenue_monday = 120 * 35
revenue_tuesday = 95 * 35
print(revenue_monday, revenue_tuesday)

# З функцією: логіка в одному місці
def revenue(orders, avg_check):
    return orders * avg_check

print(revenue(120, 35), revenue(95, 35))
````

## 1.2 Синтаксис: `def`, параметри, `return`

```
def function_name(param1, param2):    # заголовок закінчується двокрапкою
    """Що робить функція."""           # docstring — опис (бажано)
    result = ...                       # тіло — з відступом 4 пробіли
    return result                      # значення, яке отримає той, хто викликав
```

- **Параметри** — змінні в заголовку функції; **аргументи** — конкретні значення, які передаємо під час виклику.
- `return` повертає значення і **одразу завершує** функцію — код після нього не виконується.
- Функцію треба оголосити (запустити клітинку з `def`) до того, як її викликати.
- Імена функцій пишуть малими літерами через `_` і так, щоб було зрозуміло, що вони роблять: `calc_margin`, `average_check`.

````python
def average_check(revenue, orders):
    """Повертає середній чек. Якщо замовлень немає — 0."""
    if orders == 0:
        return 0              # ранній вихід: далі код не виконується
    return revenue / orders

print(average_check(42000, 120))
print(average_check(0, 0))
````

## 1.3 `return` чи `print`?

`print()` лише **показує** значення на екрані, а `return` **віддає** його коду, який викликав функцію, — і з ним можна рахувати далі. Функція без `return` повертає спеціальне значення `None` («нічого»).

````python
def total_print(a, b):
    print(a + b)          # тільки показує

def total_return(a, b):
    return a + b          # віддає значення

x = total_print(2, 3)
y = total_return(2, 3)
print("x =", x)           # None — з цим далі нічого не порахуєш
print("y * 10 =", y * 10)
````

Функція може повернути **кілька значень** — Python пакує їх у кортеж, який зручно одразу «розпакувати» в окремі змінні:

````python
def min_and_max(values):
    return min(values), max(values)

low, high = min_and_max([310, 120, 480, 250])
print("мінімум:", low, "| максимум:", high)
````

## 1.4 Docstring — документація функції

**Docstring** — рядок у потрійних лапках одразу після заголовка. Він пояснює, що робить функція, які параметри приймає й що повертає. Його показують `help()`, атрибут `__doc__` і підказки Colab, коли наводиш курсор на ім'я функції.

````python
def margin(revenue, cost):
    """Рахує маржу у відсотках.

    revenue — виручка, cost — собівартість.
    Повертає (revenue - cost) / revenue * 100.
    """
    return (revenue - cost) / revenue * 100

help(margin)
print(round(margin(1000, 650), 1), "%")
````

## 1.5 Область видимості: локальні й глобальні змінні

- **Локальні** змінні створюються всередині функції й існують лише під час її виконання.
- **Глобальні** змінні оголошені поза функціями; функція може їх **читати**.
- Python шукає ім'я за правилом **LEGB**: *Local* (функція) → *Enclosing* (зовнішня функція) → *Global* (ноутбук або модуль) → *Built-in* (вбудовані: `len`, `print`…).

````python
tax_rate = 0.2                    # глобальна змінна

def price_with_tax(price):
    tax = price * tax_rate        # tax — локальна; tax_rate читаємо з глобальної області
    return price + tax

print(price_with_tax(100))

try:
    print(tax)
except NameError as error:
    print("Помилка:", error)      # поза функцією tax не існує
````

⚠️ **Пастка:** якщо всередині функції є *присвоєння* змінній, Python вважає її локальною в усій функції. Ключове слово `global` дозволяє змінити глобальну змінну, але такий код важко читати й тестувати. Кращий шлях: **дані входять через параметри, результат виходить через `return`**.

````python
counter = 0

def increment_bad():
    counter = counter + 1         # counter тут локальна — і ще не має значення

try:
    increment_bad()
except UnboundLocalError as error:
    print("Помилка:", error)

def increment(value):             # «чиста» функція: вхід → параметр, вихід → return
    return value + 1

counter = increment(counter)
print("counter =", counter)
````

## 1.6 Виклик і комбінування функцій

Велику задачу розбивають на маленькі функції, кожна з яких робить **одну** річ, а потім комбінують: одна функція викликає іншу, а результат однієї можна одразу передати в іншу.

````python
def subtotal(price, quantity):
    return price * quantity

def apply_discount(amount, percent):
    return amount * (1 - percent / 100)

def final_price(price, quantity, discount_percent):
    amount = subtotal(price, quantity)               # функція викликає функцію
    return apply_discount(amount, discount_percent)

print(final_price(250, 4, 10))
print(round(apply_discount(subtotal(99.9, 3), 5), 2))   # вкладений виклик
````

### 🏁 Завдання 1.10 · Міні-проєкт: функції для метрик

Ти аналітик інтернет-магазину. Є продажі за 10 днів (у клітинці нижче — `DAILY_SALES`). Напиши функції **без модуля `statistics`** — сьогодні вчимося робити це самі:

| Функція | Що повертає |
|---|---|
| `mean(values)` | середнє арифметичне |
| `median(values)` | медіану: посортуй **копію** списку; непарна кількість — середній елемент, парна — середнє двох центральних |
| `value_range(values)` | розмах: максимум мінус мінімум |
| `describe(values)` | словник `{"count", "mean", "median", "range"}`, зібраний з трьох функцій вище; середнє — округлене до 2 знаків |

Вимоги:
- для порожнього списку `mean`, `median`, `value_range` повертають `None`, а `describe` — `{"count": 0, "mean": None, "median": None, "range": None}`;
