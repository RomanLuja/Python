## Блок 2. Довжина (норма) вектора і відстань

**Нагадування:**
- **Довжина (норма):** $\|\mathbf{v}\| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2}$
- **Відстань між векторами** = довжина їхньої різниці: $\|\mathbf{a} - \mathbf{b}\|$. Що менша відстань, то схожіші об'єкти.
- **Одиничний вектор** = вектор ÷ його довжина (довжина стає 1).

| Код | Що робить |
|---|---|
| `v ** 2` | кожен елемент у квадраті |
| `np.sqrt(x)` | квадратний корінь |
| `np.linalg.norm(v)` | довжина вектора |
| `np.linalg.norm(A - v, axis=1)` | відстань від `v` до **кожного рядка** матриці A |

## Блок 5. Скалярний добуток

**Нагадування:**
$$\mathbf{a} \cdot \mathbf{b} = a_1 b_1 + a_2 b_2 + \dots + a_n b_n$$

Попарно перемножуємо і складаємо → **одне число**. Це будь-яка «зважена сума»: чек = кількості · ціни, зважений бал = оцінки · ваги.

| Код | Що робить |
|---|---|
| `(a * b).sum()` | вручну |
| `np.dot(a, b)` або `a @ b` | скалярний добуток |

**Косинусна схожість:** $\cos = \dfrac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{a}\| \cdot \|\mathbf{b}\|}$ — що ближче до 1, то схожіші вектори.

## Блок 6. Множення матриць і транспонування

**Нагадування:**
- **Матриця × вектор:** кожен рядок матриці скалярно множиться на вектор → вектор результатів.
- **Матриця × матриця:** елемент результату (i, j) = рядок i першої матриці · стовпець j другої.
- **Правило розмірів:** (m × **n**) @ (**n** × p) = (m × p). Внутрішні числа мають збігатися.
- **Порядок важливий:** A @ B ≠ B @ A.
- ⚠️ `A * B` — **поелементно**, `A @ B` — **матрично**.
- **Транспонування** `A.T`: рядки стають стовпцями, розмір m × n → n × m.
- `.argmax()` повертає **номер** найбільшого елемента.
