## 1. Цикл `for` і `range()`, `enumerate()`, `zip()` — 14 завдань

### 📘 Теорія: як працює `for`

> 📌 **Шаблон синтаксису** — це схема, а не код для запуску
>
> <code>for element in collection:<br>&nbsp;&nbsp;&nbsp;&nbsp;# тіло циклу: виконується для кожного елемента</code>

На кожній ітерації змінна `element` по черзі отримує значення наступного елемента колекції. Коли елементи закінчуються, цикл завершується сам — лічильник і умова не потрібні, тому нескінченного циклу тут випадково не отримаєш.

`for` перебирає будь-яку колекцію: список, кортеж, рядок (по символах), множину (у довільному порядку), словник.

> 💡 Назву змінної циклу обираєте ви. Давайте змістовні назви: `for price in prices`, `for city in cities`, `for letter in word`.

```python
prices = [120, 85, 340, 99]
for price in prices:
    print("Price:", price)

word = "data"
for letter in word:
    print(letter.upper(), end=" ")     # D A T A
```

### 📘 Перебір словника

| Запис | Що отримує змінна циклу |
|---|---|
| `for key in d:` | ключі (так само, як `for key in d.keys():`) |
| `for value in d.values():` | значення |
| `for key, value in d.items():` | пару «ключ, значення» — кортеж одразу розпаковується у дві змінні |

```python
stock = {"apples": 40, "pears": 15, "plums": 25}

for product in stock:
    print(product)                      # apples, pears, plums

for qty in stock.values():
    print(qty)                          # 40, 15, 25

for product, qty in stock.items():
    print(product, "->", qty)           # apples -> 40 ...
```

### 📘 Накопичення результату в циклі

Більшість задач з циклами будуються за однією схемою:

1. **До циклу** створюємо «накопичувач»: `total = 0`, `count = 0`, `result = []`, `counts = {}`.
2. **У циклі** оновлюємо його для кожного елемента (іноді — лише за умови).
3. **Після циклу** використовуємо результат.

> ⚠️ Якщо створити накопичувач **усередині** циклу, він обнулятиметься на кожній ітерації, і в результаті залишиться лише значення останнього елемента.

```python
prices = [120, 85, 340, 99, 560]

total = 0          # накопичувач для суми
expensive = []     # накопичувач для відфільтрованих значень

for price in prices:
    total += price
    if price > 100:
        expensive.append(price)

print("Total:", total)            # Total: 1204
print("Expensive:", expensive)    # Expensive: [120, 340, 560]
```

### 📘 `range()`

`range()` генерує послідовність цілих чисел. Як і у зрізах, **кінцеве значення не включається**.

| Запис | Числа |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 20, 5)` | 0, 5, 10, 15 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 |

Для чого використовують:
- повторити дію певну кількість разів: `for _ in range(3):` — змінна `_` означає «значення мені не потрібне»;
- перебрати числа з певним кроком;
- перебрати **індекси** списку: `for i in range(len(lst)):` — коли поруч з елементом потрібен його номер (хоча для цього є зручніший `enumerate()`, див. нижче).

`break` і `continue` у циклі `for` працюють так само, як у `while`.

```python
for i in range(1, 6):
    print(i, end=" ")          # 1 2 3 4 5
print()                        # порожній print() — перехід на новий рядок

for i in range(10, 0, -3):
    print(i, end=" ")          # 10 7 4 1
print()

for _ in range(3):
    print("Hello!")            # тричі Hello!

cities = ["Kyiv", "Lviv", "Odesa"]
for i in range(len(cities)):
    print(i, cities[i])        # 0 Kyiv / 1 Lviv / 2 Odesa
```

### 📘 `enumerate()`: елемент разом з індексом

`enumerate(collection)` на кожній ітерації дає пару **(індекс, елемент)**. Параметр `start` задає, з якого числа починати нумерацію — зручно для нумерації «по-людськи», з 1.

```python
cities = ["Kyiv", "Lviv", "Odesa"]

for i, city in enumerate(cities):
    print(i, city)                        # 0 Kyiv / 1 Lviv / 2 Odesa

for number, city in enumerate(cities, start=1):
    print(str(number) + ".", city)        # 1. Kyiv / 2. Lviv / 3. Odesa
```

### 📘 `zip()`: паралельний перебір кількох списків

`zip(list_1, list_2, ...)` на кожній ітерації бере **по одному елементу з кожного списку** з однаковою позицією. Це потрібно, коли дані про один обʼєкт зберігаються в кількох паралельних списках: у першому — назви, у другому — ціни.

> ⚠️ `zip()` зупиняється на **найкоротшому** списку. Якщо списки різної довжини, «зайві» елементи довшого списку будуть мовчки проігноровані.

Корисний прийом: `dict(zip(keys, values))` створює словник з двох списків.

```python
products = ["laptop", "mouse", "monitor"]
prices = [32000, 650, 7800]

for product, price in zip(products, prices):
    print(product, "costs", price)       # laptop costs 32000 ...

catalog = dict(zip(products, prices))
print(catalog)       # {'laptop': 32000, 'mouse': 650, 'monitor': 7800}

short = [1, 2]
print(list(zip(products, short)))        # [('laptop', 1), ('mouse', 2)] — monitor загубився
```

### ⚠️ Не змінюйте список, який перебираєте

Якщо видаляти елементи зі списку прямо під час його перебору, елементи «зсуваються», і цикл перестрибує деякі з них. Правильний підхід — **створити новий список** лише з потрібних елементів.

```python
nums = [1, 2, 2, 3, 4]

# ❌ видаляємо під час перебору
for n in nums:
    if n == 2:
        nums.remove(n)
print(nums)        # [1, 2, 3, 4] — одна двійка залишилася!

# ✅ будуємо новий список
nums = [1, 2, 2, 3, 4]
result = []
for n in nums:
    if n != 2:
        result.append(n)
print(result)      # [1, 3, 4]
```

## 2. Вкладені цикли та типові алгоритми — 14 завдань

### 📘 Теорія: вкладені цикли

Усередині тіла циклу можна написати ще один цикл:

> 📌 **Шаблон синтаксису** — це схема, а не код для запуску
>
> <code>for outer in collection_1:<br>&nbsp;&nbsp;&nbsp;&nbsp;for inner in collection_2:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;# виконується для кожної пари (outer, inner)</code>

**Як це працює:** для **кожної** ітерації зовнішнього циклу внутрішній цикл виконується **повністю**, від початку до кінця. Якщо зовнішній цикл має 3 ітерації, а внутрішній — 4, тіло внутрішнього циклу виконається 3 × 4 = 12 разів.

**Де трапляються вкладені цикли:**
- обробка «таблиць» — списків списків: рядки × колонки;
- перебір усіх комбінацій: розміри × кольори, команда × команда;
- обробка вкладених структур: замовлення → товари в замовленні.

> ⚠️ `break` у внутрішньому циклі завершує **лише внутрішній** цикл — зовнішній продовжить роботу.

```python
sizes = ["S", "M", "L"]
colors = ["black", "white"]

for size in sizes:
    for color in colors:
        print(size, color)

# S black / S white / M black / M white / L black / L white — 3 × 2 = 6 комбінацій
```

**Список списків — таблиця.** Зовнішній цикл перебирає рядки, внутрішній — значення в рядку:

```python
sales = [
    [120, 95, 140],   # магазин 1: продажі за 3 дні
    [80, 110, 70],    # магазин 2
    [150, 60, 90],    # магазин 3
]

for store in sales:
    store_total = 0                  # накопичувач для КОЖНОГО магазину — всередині зовнішнього циклу
    for day_sales in store:
        store_total += day_sales
    print(store_total)               # 355 / 260 / 300
```

**Вкладені структури:** замовлення, у кожному з яких є список товарів.

```python
orders = [
    {"id": 1, "items": ["mouse", "pad"]},
    {"id": 2, "items": ["laptop"]},
    {"id": 3, "items": ["mouse", "cable", "hub"]},
]

total_items = 0
for order in orders:
    for item in order["items"]:
        total_items += 1

print("Total items:", total_items)   # Total items: 6
```

### 📘 Типові алгоритми

Більшість задач аналітика на «сирому» Python зводяться до кількох шаблонів. Усі вони працюють за схемою «накопичувач до циклу → оновлення в циклі → результат після циклу».

| Алгоритм | Накопичувач до циклу | Дія в циклі | Приклад задачі |
|---|---|---|---|
| **Підрахунок** | `count = 0` | якщо умова — `count += 1` | скільки замовлень дорожчі за 1000 грн |
| **Підсумовування** | `total = 0` | `total += value` | загальна виручка |
| **Фільтрація** | `result = []` | якщо умова — `result.append(value)` | список дрібних замовлень |
| **Мінімум / максимум** | `best = перший елемент` | якщо `value > best` — `best = value` | найдорожчий товар |
| **Підрахунок частот** | `counts = {}` | `counts[key] = counts.get(key, 0) + 1` | скільки разів трапляється кожна категорія |
| **Групування з сумою** | `totals = {}` | `totals[key] = totals.get(key, 0) + value` | виручка за кожною категорією |

> 💡 У Python є готові `sum()`, `max()`, `min()`, `len()`. Але алгоритми важливо вміти писати самостійно: готової функції часто немає — наприклад, коли потрібен максимум з умовою або разом з назвою товару.

**Підрахунок, підсумовування, фільтрація** — часто в одному циклі:

```python
amounts = [1200, 450, 3100, 800, 2500, 150, 990]

total = 0
count_big = 0
small_orders = []

for amount in amounts:
    total += amount                   # підсумовування
    if amount >= 1000:
        count_big += 1                # підрахунок
    if amount < 500:
        small_orders.append(amount)   # фільтрація

print("Total:", total)                # Total: 9190
print("Big orders:", count_big)       # Big orders: 3
print("Small orders:", small_orders)  # Small orders: [450, 150]
```

**Пошук мінімуму й максимуму.** Початкове значення — **перший елемент** колекції, а не 0.

> ⚠️ Якщо почати з `max_value = 0`, а всі значення відʼємні (наприклад, температури взимку), то жодне не буде більшим за 0 — і результат буде неправильним.

```python
temps = [3, -2, 5, -7, 0, 8, -1]

max_temp = temps[0]
min_temp = temps[0]
max_day = 0

for i, t in enumerate(temps):
    if t > max_temp:
        max_temp = t
        max_day = i               # запамʼятовуємо не лише значення, а й позицію
    if t < min_temp:
        min_temp = t

print("Max:", max_temp, "on day index", max_day)   # Max: 8 on day index 5
print("Min:", min_temp)                            # Min: -7

prices = {"laptop": 32000, "mouse": 650, "monitor": 7800}
best_product = ""
best_price = 0                    # тут 0 підходить: ціни завжди додатні
for product, price in prices.items():
    if price > best_price:
        best_price = price
        best_product = product
print(best_product, best_price)   # laptop 32000
```

**Підрахунок частот і групування.** Метод `get(key, 0)` повертає 0, якщо ключа ще немає в словнику, тому той самий рядок коду працює і для першої появи ключа, і для наступних.

```python
categories = ["food", "tech", "food", "books", "tech", "food"]

counts = {}
for category in categories:
    counts[category] = counts.get(category, 0) + 1
print(counts)        # {'food': 3, 'tech': 2, 'books': 1}

# групування з сумою: виручка за категоріями
amounts = [120, 3000, 80, 450, 1500, 200]
revenue = {}
for category, amount in zip(categories, amounts):
    revenue[category] = revenue.get(category, 0) + amount
print(revenue)       # {'food': 400, 'tech': 4500, 'books': 450}
```

### 📘 Теорія: що таке генератор списку

**Генератор списку** (англ. *list comprehension*) — короткий запис для типової ситуації «створити новий список, перебравши іншу колекцію». Те, що в циклі займає 3–4 рядки, генератор записує одним виразом.

> 📌 **Шаблон синтаксису** — це схема, а не код для запуску
>
> <code>new_list = [expression for item in iterable]</code>

Читається майже як звичайне речення: «список з **виразу** для кожного **елемента** в **колекції**».

| Частина | Що означає | Приклад |
|---|---|---|
| `expression` | що потрапить у новий список | `price * 1.2`, `name.upper()` |
| `item` | змінна, що по черзі отримує елементи | `price`, `name` |
| `iterable` | що перебираємо | список, рядок, `range()`, `d.items()` |

**Той самий результат — циклом і генератором:**

```python
prices = [100, 250, 80]

# цикл
with_vat = []
for price in prices:
    with_vat.append(price * 1.2)
print(with_vat)          # [120.0, 300.0, 96.0]

# генератор списку
with_vat = [price * 1.2 for price in prices]
print(with_vat)          # [120.0, 300.0, 96.0]

squares = [n ** 2 for n in range(1, 6)]
print(squares)           # [1, 4, 9, 16, 25]

letters = [ch.upper() for ch in "data"]
print(letters)           # ['D', 'A', 'T', 'A']
```

### 📘 Умови в генераторах: фільтрація та вибір значення

Умову можна поставити у **двох різних місцях**, і вони роблять різні речі:

| Де стоїть умова | Шаблон | Що робить | Довжина результату |
|---|---|---|---|
| **В кінці** — фільтр | `[x for x in data if condition]` | бере лише елементи, для яких умова істинна | така сама або менша |
| **На початку** — тернарний оператор | `[a if condition else b for x in data]` | для **кожного** елемента обирає одне з двох значень | завжди така сама |

> ⚠️ Фільтр у кінці пишеться **без** `else`. Тернарний оператор на початку **обовʼязково** має `else`. Ці дві форми можна поєднувати, але такий код уже важче читати.

```python
amounts = [1200, 450, 3100, 800, 150]

# фільтр: лише великі замовлення
big = [a for a in amounts if a >= 1000]
print(big)          # [1200, 3100]

# тернарний оператор: мітка для кожного замовлення
labels = ["big" if a >= 1000 else "small" for a in amounts]
print(labels)       # ['big', 'small', 'big', 'small', 'small']

# фільтр + перетворення: знижка 10% лише для великих замовлень
discounted = [a * 0.9 for a in amounts if a >= 1000]
print(discounted)   # [1080.0, 2790.0]
```

### 📘 Генератори словників і множин

Той самий принцип працює для словників і множин — змінюються лише дужки:

| Що створюємо | Шаблон |
|---|---|
| Список | `[expression for item in iterable]` |
| Множину | `{expression for item in iterable}` |
| Словник | `{key: value for item in iterable}` |

Множину від словника Python відрізняє за **двокрапкою**: є `key: value` — це словник, немає — множина. Фільтр `if` у кінці працює в усіх трьох видах генераторів.

```python
products = ["laptop", "mouse", "monitor"]
prices = [32000, 650, 7800]

# словник з двох списків
catalog = {product: price for product, price in zip(products, prices)}
print(catalog)        # {'laptop': 32000, 'mouse': 650, 'monitor': 7800}

# фільтрація словника: лише дорогі товари
expensive = {p: price for p, price in catalog.items() if price > 1000}
print(expensive)      # {'laptop': 32000, 'monitor': 7800}

# перетворення значень: ціни в доларах
in_usd = {p: round(price / 41.2, 2) for p, price in catalog.items()}
print(in_usd)         # {'laptop': 776.7, 'mouse': 15.78, 'monitor': 189.32}

# множина: унікальні перші літери назв
first_letters = {p[0] for p in products}
print(sorted(first_letters))   # ['l', 'm']
```

### 📘 Вкладені цикли в генераторі

Генератор може містити кілька `for`. Вони записуються **в тому самому порядку**, що й вкладені цикли: спочатку зовнішній, потім внутрішній. Найчастіше так «розгортають» список списків в один плаский список.

> 💡 Для допитливих: якщо замість квадратних дужок поставити круглі, вийде **генераторний вираз**. Його зручно передавати прямо у функції: `sum(price for price in prices if price > 100)` — Python не створюватиме проміжний список.

```python
sales = [[120, 95], [80, 110], [150, 60]]

# вкладені цикли
flat = []
for row in sales:
    for value in row:
        flat.append(value)
print(flat)            # [120, 95, 80, 110, 150, 60]

# той самий порядок for — у генераторі
flat = [value for row in sales for value in row]
print(flat)            # [120, 95, 80, 110, 150, 60]

print(sum(value for value in flat if value > 100))   # 380 — генераторний вираз
```

### 📘 Як перетворити цикл на генератор

Генератором можна замінити цикл, який має саме таку будову:

| Частина циклу | Куди переходить у генераторі |
|---|---|
| `result = []` перед циклом | квадратні дужки навколо генератора |
| `for x in data:` | `for x in data` всередині дужок |
| `if condition:` перед `append` | `if condition` **в кінці** |
| `result.append(вираз)` | **вираз на початку** генератора |

**Генератором НЕ замінюють цикл, якщо в ньому:**
- є `break` або `continue` зі складною логікою;
- оновлюються **кілька** накопичувачів одночасно (сума, лічильник і список разом);
- виконуються дії «заради ефекту»: `print()`, `input()`, зміна інших змінних;
- кілька кроків обчислення, які в один вираз не вміщаються без втрати зрозумілості.

```python
emails = ["Anna@Gmail.com", "bohdan@ukr.net", "IRYNA@gmail.com"]

# цикл
gmail_users = []
for email in emails:
    if email.lower().endswith("@gmail.com"):
        gmail_users.append(email.lower())

# генератор: вираз з append → на початок, умова з if → у кінець
gmail_users = [email.lower() for email in emails if email.lower().endswith("@gmail.com")]
print(gmail_users)     # ['anna@gmail.com', 'iryna@gmail.com']
```

### 📘 Читабельність коду

Генератор має робити код **зрозумілішим**, а не коротшим будь-якою ціною. Код читають значно частіше, ніж пишуть.

| ✅ Добре | ❌ Погано |
|---|---|
| один `for` і, можливо, один `if` | три вкладені `for` з умовами в одному рядку |
| рядок до ~80 символів | рядок, який доводиться прокручувати вбік |
| змістовні імена: `price for price in prices` | `x for x in y` у складній логіці |
| результат генератора кудись записується | `[print(x) for x in data]` — генератор лише заради побічного ефекту |
| простий тернарний оператор | тернарний оператор усередині тернарного оператора |

Довгий генератор можна розбити на кілька рядків усередині дужок — Python це дозволяє. А якщо логіка не вміщається навіть так — пишіть звичайний цикл, це не помилка.

```python
orders = [
    {"id": 1, "amount": 1200, "status": "paid"},
    {"id": 2, "amount": 450, "status": "new"},
    {"id": 3, "amount": 3100, "status": "paid"},
]

# ❌ важко читати
ids = [o["id"] for o in orders if o["status"] == "paid" and o["amount"] > 1000]

# ✅ той самий генератор, розбитий на рядки
paid_big_ids = [
    order["id"]
    for order in orders
    if order["status"] == "paid" and order["amount"] > 1000
]
print(paid_big_ids)    # [1, 3]
```

### 📘 Теорія: поєднання умов і циклів

Реальні програми майже завжди поєднують цикли й умови. Більшість з них складаються з кількох повторюваних шаблонів:

| Шаблон | Будова | Приклад |
|---|---|---|
| **Обробка кожного елемента з умовою** | `for` + `if` / `elif` / `else` всередині | класифікувати кожну транзакцію |
| **Пошук до першого збігу** | `for` + `if` + `break` | знайти перше замовлення понад 10 000 грн |
| **Повторювати до потрібного результату** | `while` + умова | рахувати відсотки, доки сума не досягне цілі |
| **Перевірка введення** | `while True` + `if` + `break` | питати число, доки не введуть правильне |
| **Меню програми** | `while True` + `match` або `if` / `elif` + `break` для виходу | консольний застосунок з пунктами меню |

**Меню консольної програми** — це головний цикл, який на кожній ітерації:
1. показує пункти меню;
2. читає вибір користувача;
3. виконує відповідну дію;
4. повертається до пункту 1 — доки користувач не обере вихід.

```python
shopping_list = []

while True:
    print("1 - add item, 2 - show list, 0 - exit")
    choice = input("Your choice: ")

    match choice:
        case "1":
            item = input("Item name: ").strip()
            if item:
                shopping_list.append(item)
                print("Added:", item)
            else:
                print("Item name cannot be empty")
        case "2":
            if shopping_list:
                for number, item in enumerate(shopping_list, start=1):
                    print(str(number) + ".", item)
            else:
                print("List is empty")
        case "0":
            print("Goodbye!")
            break
        case _:
            print("Unknown option")
```

### 📘 Перевірка числа, введеного як рядок

`isdigit()` повертає `True` лише для **невідʼємних цілих** чисел: для `"-40"` чи `"98.6"` результат буде `False`, хоча це цілком коректні числа.

Щоб перевірити відʼємне або дробове число, перед перевіркою з рядка прибирають один мінус на початку (`removeprefix("-")`) і одну крапку (`replace(".", "", 1)`). Якщо після цього залишилися лише цифри — рядок можна безпечно перетворити на число через `float()`.

```python
for text in ["25", "-40", "98.6", "3.1.4", "abc", ""]:
    digits = text.removeprefix("-").replace(".", "", 1)
    print(repr(text), digits.isdigit())

# '25' True / '-40' True / '98.6' True / '3.1.4' False / 'abc' False / '' False

value = "98.6"
if value.removeprefix("-").replace(".", "", 1).isdigit():
    number = float(value)
    print(number + 1)      # 99.6
```

### 📘 Псевдокод

**Псевдокод** — опис алгоритму людською мовою, але зі структурою програми: кроки, умови, цикли, відступи. Він не має суворих правил синтаксису і не запускається. Його мета — продумати логіку, перш ніж відволікатися на деталі Python.

> 📝 **Псевдокод — схема рішення, а не код для запуску**
>
> сума = 0, кількість = 0, найбільше = немає<br>ДЛЯ КОЖНОГО замовлення В списку замовлень:<br>&nbsp;&nbsp;&nbsp;&nbsp;ЯКЩО статус замовлення = &quot;оплачено&quot;:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;ЗБІЛЬШИТИ суму на суму замовлення<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;ЗБІЛЬШИТИ кількість на 1<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;ЯКЩО найбільше = немає АБО замовлення більше за найбільше:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;найбільше = це замовлення<br>ЯКЩО кількість &gt; 0:<br>&nbsp;&nbsp;&nbsp;&nbsp;ВИВЕСТИ сума / кількість та найбільше<br>ІНАКШЕ:<br>&nbsp;&nbsp;&nbsp;&nbsp;ВИВЕСТИ &quot;немає оплачених замовлень&quot;

**Той самий алгоритм на Python** — кожен рядок псевдокоду перетворився на один-два рядки коду:

```python
orders = [
    {"id": 1, "amount": 1200, "status": "paid"},
    {"id": 2, "amount": 450, "status": "new"},
    {"id": 3, "amount": 3100, "status": "paid"},
    {"id": 4, "amount": 800, "status": "paid"},
]

total = 0
count = 0
largest = None              # None — «значення ще немає»

for order in orders:
    if order["status"] == "paid":
        total += order["amount"]
        count += 1
        if largest is None or order["amount"] > largest["amount"]:
            largest = order

if count > 0:
    print("Average paid order:", round(total / count, 2))   # Average paid order: 1700.0
    print("Largest paid order:", largest["id"], largest["amount"])   # Largest paid order: 3 3100
else:
    print("No paid orders")
```

> 💡 `None` — спеціальне значення «нічого немає». Порівнювати з ним прийнято через `is`: `x is None`. Це зручне початкове значення для пошуку максимуму, коли першого елемента ще немає або його треба спершу перевірити умовою.

## 🧭 Підсумок

| Ситуація | Що використати |
|---|---|
| Перебрати всі елементи колекції | `for element in collection` |
| Повторити дію N разів або перебрати числа | `for i in range(...)` |
| Потрібні і елемент, і його номер | `enumerate()` |
| Перебрати кілька паралельних списків | `zip()` |
| Обробити «таблицю», комбінації або вкладену структуру | вкладені цикли |
| Підрахунок, сума, фільтрація, мінімум/максимум, частоти | накопичувач до циклу → оновлення в циклі |
| Створити нову колекцію з іншої простим перетворенням чи фільтром | генератор списку, словника або множини |
| Кілька накопичувачів, `break`, `print()`, складна логіка | звичайний цикл |
| Програма з меню, яка працює до команди виходу | `while True` + `match` / `if` + `break` |
| Складна задача | план по кроках → псевдокод → код по одному кроку |

**Чек-лист:**
- накопичувач (`total`, `count`, `result`, `counts`) створено **до** циклу;
- для пошуку мінімуму чи максимуму початкове значення — перший елемент або `None`, а не 0;
- список не змінюється під час його перебору;
- у генераторі фільтр `if` стоїть у кінці, а вибір значення `... if ... else ...` — на початку;
- генератор читається з першого разу; якщо ні — краще звичайний цикл;
- кожен крок програми перевірено на невеликих даних, які легко порахувати вручну.
