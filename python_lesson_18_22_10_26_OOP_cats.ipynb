# Частина 1. Що таке клас і об'єкт

## Ви вже користувалися об'єктами!

У Python усе — об'єкти. Рядок `"coffee"` — це об'єкт, а список `[100, 200]` — теж об'єкт. А коли Ви писали `"coffee".upper()` чи `sales.append(300)`, Ви викликали **дії самого об'єкта**.

```python
text = "coffee"
print(type(text))       # <class 'str'> — рядок є об'єктом класу str
print(text.upper())     # upper — дія, яку вміє робити кожен рядок

sales = [100, 200]
print(type(sales))      # <class 'list'> — список є об'єктом класу list
sales.append(300)       # append — дія, яку вміє робити кожен список
print(sales)
```

Тепер ми навчимося створювати **власні** класи — наприклад, клас «Котик».

## Клас і об'єкт: у чому різниця

Уявіть притулок для котиків. Для кожного нового котика заповнюють **анкету**.

- **Клас** — це **бланк анкети**: у ньому описано, *які* поля є в кожного котика (ім'я, вік, колір) і *що* котик уміє (нявкати, спати, їсти).
- **Об'єкт** — це **заповнена анкета** конкретного котика: Felix, 3 роки, рудий.

> <code>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;клас Cat  (бланк анкети)<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;┌──────────────┼──────────────┐<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;▼&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; ▼&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; ▼<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;🐱 Felix &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;      🐱 Luna        &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;🐱 Balu  &nbsp;&nbsp;&nbsp;    ← об'єкти (конкретні котики)<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;3 р., рудий    &nbsp;&nbsp;&nbsp;&nbsp;1 р., біла     &nbsp;&nbsp;&nbsp;7 р., сірий</code>

Один клас — скільки завгодно об'єктів. Усі котики мають **однакові поля**, але **різні значення**.

| | Клас | Об'єкт |
|---|---|---|
| Що це | опис, «креслення» | конкретний екземпляр |
| Приклад | `Cat` — котик узагалі | Felix, 3 роки, рудий |
| Скільки | один | скільки завгодно |

**Два головні слова ООП:**
- **атрибути** — **дані** об'єкта (що котик *має*): ім'я, вік, колір, вага;
- **методи** — **дії** об'єкта (що котик *уміє*): нявкати, їсти, спати.

# Частина 2. Перший клас `Cat`

## 2.1 Найпростіший клас

Клас оголошують словом `class`. Назву класу прийнято писати **з великої літери**: `Cat`, `Shelter`.

```python
class Cat:
    pass          # pass означає «поки що порожньо»


felix = Cat()   # створюємо об'єкт: назва класу + дужки
balu = Cat()    # ще один, окремий котик

print(type(felix))
print(felix)
```

Що ми бачимо:
- `type(felix)` → `<class '__main__.Cat'>` — `felix` є об'єктом класу `Cat`;
- `print(felix)` показує щось на кшталт `<__main__.Cat object at 0x7f...>` — це «адреса» котика в пам'яті. Не лякайтеся: у частині 5 ми навчимо котика представлятися гарно.

## 2.2 `__init__`: заповнюємо анкету

Порожній котик без імені нам не потрібен. Щоб під час створення котика одразу заповнити його анкету, у клас додають спеціальний метод `__init__` (від англ. *initialize* — «налаштувати на початку»).

```python
class Cat:
    def __init__(self, name, age, color):
        self.name = name
        self.age = age
        self.color = color


felix = Cat("Felix", 3, "ginger")
luna = Cat("Luna", 1, "white")

print(felix.name, felix.age, felix.color)
print(luna.name, luna.age, luna.color)
```

**Що відбувається під час `Cat("Felix", 3, "ginger")` — покроково:**

1. Python створює нового, ще «порожнього» котика.
2. **Автоматично** викликає `__init__`, і в нього потрапляють: `self` = цей новий котик, `name = "Felix"`, `age = 3`, `color = "ginger"`.
3. Рядок `self.name = name` записує ім'я **в анкету саме цього котика**. Так само для віку й кольору.
4. Готовий котик потрапляє в змінну `felix`.

**Що таке `self`?** Це слово означає «**цей самий котик**», з яким зараз працюємо. `self.name` — «ім'я цього котика». Під час створення котика `self` у дужках **не пишемо**: Python підставляє його сам.

**Як прочитати атрибут:** назва об'єкта, крапка, назва атрибута — `felix.name`.

💡 Подвійні підкреслення з обох боків `__init__` позначають «спеціальний» метод, який Python викликає сам. Такі методи ще називають «магічними».

## 2.3 Атрибути можна змінювати

Атрибут об'єкта змінюється так само, як звичайна змінна. Змінюється він **лише в цього котика**, а інші котики залишаються без змін.

```python
luna.age = 2           # у Luna день народження: тепер їй 2 роки
print(luna.age)
print(felix.age)        # у Felix вік не змінився
```

## ⚠️ Типові помилки

| Що бачите | Причина | Як виправити |
|---|---|---|
| `TypeError: Cat.__init__() missing 1 required positional argument: 'color'` | під час створення котика передали не всі дані | `Cat("Felix", 3, "ginger")` — усі три значення |
| `TypeError: Cat.__init__() takes 4 positional arguments but 5 were given` | передали зайве значення | перевірте, скільки параметрів у `__init__` |
| `AttributeError: 'Cat' object has no attribute 'nme'` | помилка в назві атрибута | `felix.name` |
| `AttributeError: 'NoneType' object has no attribute 'name'` | котика ще не створено: у змінній досі `None` | замініть `None` на `Cat(...)` |
| атрибут не зберігся | у `__init__` написали `name = name` замість `self.name = name` | без `self.` значення не потрапляє в анкету котика |
| `TypeError: Cat.__init__() takes 3 positional arguments but 4 were given` | забули `self` першим параметром у `__init__`: Python усе одно передає котика першим аргументом, тож аргументів виявляється на один більше | `def __init__(self, name, age, color):` |

# Частина 3. Методи: що котик уміє робити

## 3.1 Метод — це функція всередині класу

Атрибути — це **дані** котика. А **методи** — це **дії**: нявкати, спати, гратися. Метод пишеться як звичайна функція (`def`), але:
- він стоїть **всередині класу** (з відступом);
- його **перший параметр — завжди `self`**, тобто «цей котик»;
- викликають його **через крапку**: `felix.sleep()`.

```python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.naps = 0     # не з параметра: у кожного нового котика спочатку 0 снів

    def sleep(self):
        self.naps += 1
        return f"{self.name} is sleeping 😴 (nap #{self.naps} today)"

    def play(self, toy):
        return f"{self.name} is playing with {toy}"


felix = Cat("Felix", 3)
luna = Cat("Luna", 1)

print(felix.sleep())
print(felix.sleep())
print(luna.sleep())
print(felix.play("a ball"))
```

**Що тут відбувається:**
1. `felix.sleep()` — Python викликає метод `sleep`, підставляючи `self = felix`.
2. `self.naps += 1` збільшує лічильник снів **саме котика Felix**.
3. У Luna свій лічильник, тому її перший сон — `nap #1`.
4. `felix.play("a ball")` — метод з параметром: `self = felix`, `toy = "a ball"`.

Зверніть увагу на `self.naps = 0` у `__init__`: атрибут можна створити й **без параметра** — тоді в кожного нового котика буде однакове початкове значення.

| | Звичайна функція | Метод |
|---|---|---|
| Де пишемо | окремо | всередині класу |
| Перший параметр | будь-який | завжди `self` |
| Як викликаємо | `sleep(felix)` | `felix.sleep()` |
| Звідки бере дані | з параметрів | з параметрів **і** з атрибутів через `self` |

## ⚠️ Типові помилки

| Що бачите | Причина | Як виправити |
|---|---|---|
| `TypeError: Cat.sleep() takes 0 positional arguments but 1 was given` | забули `self` у методі | `def sleep(self):` |
| `NameError: name 'name' is not defined` | у методі написали `name` замість `self.name` | дані котика беремо через `self.` |
| `<bound method Cat.sleep of ...>` | забули дужки під час виклику | `felix.sleep()` |
| `AttributeError: 'Cat' object has no attribute 'meow'` | такого методу в класі немає, або забули перезапустити клітинку з класом | перевірте назву й запустіть клітинку з класом |

# Частина 4. Багато котиків: аналізуємо притулок

## 4.1 Повний клас `Cat`

Зберемо все, що ми написали, в один клас. **Запустіть клітинку нижче**: з цим класом працюватимемо далі.

```python
class Cat:
    def __init__(self, name, age, color, weight):
        self.name = name
        self.age = age
        self.color = color
        self.weight = weight

    def meow(self):
        return f"{self.name}: Meow!"

    def describe(self):
        return f"{self.name}, {self.age} y.o., {self.color}, {self.weight} kg"

    def birthday(self):
        self.age += 1
        return f"Happy birthday, {self.name}! Now {self.age} y.o."

    def eat(self, grams):
        self.weight = round(self.weight + grams / 1000, 2)
```

## 4.2 Список об'єктів — це таблиця

Котиків можна зберігати у **списку**, як звичайні числа чи рядки. Для аналітика це звична річ: список об'єктів — це **таблиця**, де кожен об'єкт — рядок, а атрибути — стовпці.

| `name` | `age` | `color` | `weight` |
|---|---|---|---|
| Felix | 3 | ginger | 4.5 |
| Luna | 1 | white | 3.2 |
| … | … | … | … |

**Запустіть клітинку з мешканцями притулку:**

```python
SHELTER_CATS = [
    Cat("Felix", 3, "ginger", 4.5),
    Cat("Luna", 1, "white", 3.2),
    Cat("Balu", 7, "grey", 5.8),
    Cat("Nalu", 2, "ginger", 4.1),
    Cat("Minka", 10, "black", 3.9),
    Cat("Milo", 4, "white", 6.3),
]

print("Cats in the shelter:", len(SHELTER_CATS))

for cat in SHELTER_CATS:       # cat — по черзі кожен котик зі списку
    print(cat.meow())
```

У циклі `for cat in SHELTER_CATS:` змінна `cat` по черзі стає кожним котиком. Тому з нею можна робити все те саме, що з `felix`: читати атрибути (`cat.age`) і викликати методи (`cat.meow()`).

# Частина 5. Гарний `print`: метод `__str__`

Пам'ятаєте, що `print(felix)` показує незрозуміле `<__main__.Cat object at 0x7f...>`? Це можна виправити.

Якщо додати в клас спеціальний метод `__str__`, то `print()` **сам** викличе його й покаже рядок, який метод повертає.

```python
class Mouse:
    def __init__(self, name):
        self.name = name


class Hamster:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"🐹 Hamster {self.name}"


print(Mouse("Pip"))       # без __str__ — «адреса» в пам'яті
print(Hamster("Nibbles"))    # з __str__ — гарний текст
```

**Правила для `__str__`:**
- назва — рівно `__str__`, по два підкреслення з кожного боку;
- єдиний параметр — `self`;
- метод **повертає** рядок (`return`), а не друкує його.

# Частина 6. Наслідування (оглядово)

## 6.1 Ідея: «кошеня — теж котик»

Буває, що потрібен **схожий** клас, у якому майже все як у `Cat`, але трохи інакше. Наприклад, кошеня: воно має ім'я, вік і вагу, як і дорослий котик, але нявкає тоненько й любить гратися.

Переписувати весь клас заново не потрібно. Можна сказати Python: «`Kitten` — це `Cat`, але з такими відмінностями». Це називається **наслідування**:

> <code>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Cat  (батьківський клас)<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;│   ім'я, вік, колір, вага, meow(), describe()<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;▼<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Kitten  (дочірній клас)<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;отримує все від Cat<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;+ своє: інший meow(), новий play()</code>

Спершу **запустіть клітинку з повним класом `Cat`**:

```python
class Cat:
    def __init__(self, name, age, color, weight):
        self.name = name
        self.age = age
        self.color = color
        self.weight = weight

    def meow(self):
        return f"{self.name}: Meow!"

    def describe(self):
        return f"{self.name}, {self.age} y.o., {self.color}, {self.weight} kg"

    def __str__(self):
        return f"🐱 {self.name} ({self.age} y.o., {self.color})"
```

## 6.2 Дочірній клас

Батьківський клас пишуть у дужках після назви: `class Kitten(Cat):`.

```python
class Kitten(Cat):
    def meow(self):                       # замінюємо метод батька своїм
        return f"{self.name}: mew-mew (very quietly)"

    def play(self):                       # новий метод, якого в Cat немає
        return f"{self.name} is chasing its own tail 🌀"


kiki = Kitten("Kiki", 0, "tabby", 0.8)  # __init__ успадковано від Cat

print(kiki)                # __str__ успадковано від Cat
print(kiki.describe())     # describe успадковано від Cat
print(kiki.meow())         # meow — свій, з класу Kitten
print(kiki.play())         # play — є лише в кошенят
```

**Що успадкував `Kitten`:**

| Що | Звідки |
|---|---|
| `__init__`, `describe`, `__str__` | **від `Cat`** — у `Kitten` ми їх не писали |
| `meow` | **свій**: замінив батьківський метод |
| `play` | **новий**: є лише в `Kitten` |

## 6.3 Один виклик — різна поведінка

Покладемо котика й кошеня в один список і попросимо кожного нявкнути. Метод той самий — `meow()`, але кожен відповідає **по-своєму**.

```python
pets = [Cat("Felix", 3, "ginger", 4.5), Kitten("Kiki", 0, "tabby", 0.8)]

for pet in pets:
    print(pet.meow())
```

Це називається **поліморфізм** (з грецької — «багато форм»): однаковий виклик, різна поведінка залежно від класу об'єкта.

# Частина 7. Підсумок і міні-проєкт

## 7.1 Чотири ідеї ООП — простими словами

Це огляд: детально кожну ідею вивчають пізніше. Ви вже зустріли їх усі!

| Ідея | Простими словами | Де Ви це бачили |
|---|---|---|
| **Інкапсуляція** | дані й дії зберігаються разом, «в одній коробці» | ім'я, вік і `meow()` — усе всередині `Cat` |
| **Наслідування** | новий клас отримує все від існуючого | `Kitten(Cat)`, `Lion(Cat)` |
| **Поліморфізм** | однаковий виклик — різна поведінка | `meow()` у котика, кошеняти й лева |
| **Абстракція** | беремо лише те, що важливо для задачі | для притулку важливі ім'я, вік, вага — а не кількість вусів |

## 7.2 Де ООП у роботі аналітика

Ви постійно користуватиметеся **чужими** класами. Наприклад, у бібліотеці `pandas` таблиця — це об'єкт класу `DataFrame`:

> <code>import pandas as pd<br><br>cats = pd.DataFrame({"name": ["Felix", "Luna"], "age": [3, 1]})   # створюємо об'єкт<br>cats["age"].mean()     # метод: середній вік<br>cats.head()            # метод: перші рядки таблиці<br>cats.shape             # атрибут: розмір таблиці</code>

Тепер Ви розумієте, що означають ці крапки: **об'єкт → його атрибути й методи**. Так само, як `felix.name` і `felix.meow()`!

## 7.3 Шпаргалка

| Що | Як записати |
|---|---|
| Оголосити клас | `class Cat:` |
| Анкета нового об'єкта | `def __init__(self, name, age):` |
| Записати атрибут | `self.name = name` |
| Створити об'єкт | `felix = Cat("Felix", 3)` |
| Прочитати атрибут | `felix.name` |
| Змінити атрибут | `felix.age = 4` |
| Метод | `def meow(self):` всередині класу |
| Викликати метод | `felix.meow()` |
| Гарний `print` | `def __str__(self): return "..."` |
| Наслідування | `class Lion(Cat):` |
